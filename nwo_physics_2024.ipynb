Ground State Sign Structures as Boolean Functions

*Abstract text (max 1500 characters)*

```python
abstract = """The ground state of a quantum system with a real-valued Hermitian Hamiltonian is a real-valued vector in the corresponding Hilbert space; thus, its phase components are reduced to signs. As recently shown [1], the sign structures of frustrated magnetic systems appear to be very complex: specifically, it is a challenging problem to reconstruct such a structure from its small part using popular neural network architectures. This complexity mitigates the success of variational methods in solving many interesting quantum systems.

Here, we study the complexity of sign structures for spin-1/2 systems using the methods of Boolean Fourier Analysis. We represent the sign structure as a mapping from the Boolean hypercube {-1, 1}^n to the set {-1, 0, 1} and write it as a polynomial with real coefficients. We then introduce several complexity measures associated with these polynomials and demonstrate that the sign structures previously known to be difficult for neural networks, specifically those corresponding to spin liquids, are also complex in the sense of these measures. We also compare the complexities of sign structures with those of amplitudes and demonstrate that, in the frustrated regime, the latter are simpler than the former.

[1] Tom Westerhout et al. https://www.nature.com/articles/s41467-020-15402-w
"""
```
