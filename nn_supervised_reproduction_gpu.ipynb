```python
from conv2d_circular import InvariantSpinCNNRegression
from spin_lattices import TriangularLattice
from heisenberg_hamiltonians import HeisenbergJ1J2
import torch
import numpy as np
```

```python
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
```

```python
lattice = TriangularLattice(6, 4)
system = HeisenbergJ1J2(lattice, J2=1.3, use_symmetries=False, spin_inversion=None)
net = InvariantSpinCNNRegression(
    lattice=lattice,
    hidden_channels=[32, 32, 32],
    dilations=[3, 2, 1],
    kernel_size=3,
    out_dim=1,
    last_layer_bias=False,
)
net.to(device)
```

```python
net(torch.from_numpy(system.basis.states[0: 2].astype(np.int64)).to(device))
```
