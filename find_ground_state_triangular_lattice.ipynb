```python
from sympy.combinatorics import Permutation, PermutationGroup
from find_ground_state_triangular_lattice_parse_symmetries import parsed_symmetries
from spin_lattices import TriangularLattice
from spin_systems import SpinSystem, heisenberg
import lattice_symmetries as ls
from pathlib import Path
```

```python
lattice = TriangularLattice(6, 6)
basis = ls.SpinBasis(lattice.number_spins, hamming_weight=lattice.number_spins // 2, spin_inversion=1, symmetries=parsed_symmetries)
hamiltonian = ls.Operator(expression=heisenberg(lattice, J2=1.3).expr, basis=basis)
system = SpinSystem(lattice, hamiltonian, ground_state_cache_dir=Path("groundstates_cache"))
system.ground_energy
```
