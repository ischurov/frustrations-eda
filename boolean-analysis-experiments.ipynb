```python
def get_all_time_top_subsets(analyzers, lat_name, top, eigenstate=0):
    subsets = set([])
    for J2 in tqdm(J2values):
        analyzer = analyzers[J2, lat_name]
        spectre_df = analyzer.get_spectre_df(
            SignalOption(eigenstate=eigenstate)
        )
        subsets.update(spectre_df.iloc[:top].index)
    return subsets
```

```python
def plot_top_state_coeffs(
    analyzers,
    lat_name,
    top,
    eigenstate=0,
    keep_first_accuracy=(1, 2, 5, 10),
    y="abs_coeff",
    plot_top_subsets=5,
):
    subsets = list(
        get_all_time_top_subsets(analyzers, lat_name, top, eigenstate)
    )
    data = []
    for J2 in J2values:
        spectre = analyzers[J2, lat_name].get_spectre_df(
            SignalOption(eigenstate=eigenstate)
        )
        data.append(
            spectre.loc[subsets]
            .reset_index()
            .rename(columns={"index": "state"})
            .assign(state=lambda x: x["state"].astype(str))
            .assign(J2=J2)
        )
    df = pd.concat(data, axis=0)

    fig = plt.figure(tight_layout=True, figsize=(13, 8))
    axd = fig.subplot_mosaic(
        [
            ["coeffs"] * plot_top_subsets,
            ["accuracy"] * plot_top_subsets,
            [f"state{i}" for i in range(plot_top_subsets)],
        ]
    )
    axd["accuracy"].sharex(axd["coeffs"])
    for i in range(1, plot_top_subsets):
        axd[f"state{i}"].sharey(axd["state0"])

    # ax1.set_yscale("log")
    sns.lineplot(data=df, x="J2", y=y, hue="state", ax=axd["coeffs"])

    accuracy_data = []
    for J2 in J2values:
        analyzer = analyzers[J2, lat_name]
        for keep_first in keep_first_accuracy:
            print(f"{J2=}, {keep_first=}, {eigenstate=}")
            accuracy = accuracy_score(
                np.sign(
                    analyzer.predict(
                        SignalOption("sign", eigenstate), keep_first=keep_first
                    )
                ),
                np.sign(
                    analyzer.system.get_df_eigenstate(
                        k=eigenstate, canonical_basis=True
                    )["eigenstate_coeff"]
                ),
            )
            accuracy_data.append(
                {"J2": J2, "accuracy": accuracy, "keep_first": str(keep_first)}
            )

    sns.lineplot(
        data=pd.DataFrame(accuracy_data),
        x="J2",
        y="accuracy",
        hue="keep_first",
        style="keep_first",
        ax=axd["accuracy"],
    )

    def onclick(event):
        J2 = event.xdata
        J2 = J2values[np.argmin(np.abs(J2values - J2))]
        analyzer = analyzers[J2, lat_name]

        multiplicities = (
            batched_state_info_df(
                analyzer.fourier_basis, analyzer.canonical_fourier_basis.states
            )
            .reset_index()
            .rename(columns={"index": "subset"})
            .groupby("representative")
            .size()
        )

        spectre = analyzer.get_spectre_df(
            signal=SignalOption(eigenstate=eigenstate)
        )

        for i in range(plot_top_subsets):
            ax = axd[f"state{i}"]
            ax.clear()

            analyzer.visualize_spectre_support_lattice(
                signal=SignalOption(eigenstate=eigenstate), m=i, ax=ax
            )
            ax.set_title(
                f"{spectre.index[i]} ({multiplicities[spectre.index[i]]})"
            )
            ax.set_ylabel(f"{J2=}")
            ax.set_aspect("equal", "box")

    cid = fig.canvas.mpl_connect("button_press_event", onclick)

    return axd
```

```python
def predict(
    self, signal: SignalOption = SignalOption(), keep_first=None
) -> np.array:
    coeffs = self.get_spectre_df(signal)["coeff"]
    if keep_first is not None:
        coeffs = coeffs.iloc[:keep_first]

    if self.use_symmetries:
        state_info_df = batched_state_info_df(
            self.fourier_basis, self.canonical_fourier_basis.states
        )
        coeffs = state_info_df.join(coeffs, on="representative")[
            "coeff"
        ].dropna()

    if keep_first is None:
        if self.canonical_fourier_transform_matrix is None:
            self.canonical_fourier_transform_matrix = get_fourier_transform_matrix(
                states=self.system.canonical_basis.states,
                subsets=self.canonical_fourier_basis.states,
                number_spins=self.system.number_spins,
                fourier_transform_matrix_cache=self.fourier_transform_matrix_cache,
                show_progress=True,
            )

        coeffs = coeffs.loc[self.canonical_fourier_basis.states]

        return self.canonical_fourier_transform_matrix @ coeffs
    else:
        transform_matrix = get_fourier_transform_matrix(
            states=self.system.canonical_basis.states,
            subsets=np.array(coeffs.index, dtype="uint64"),
            number_spins=self.system.number_spins,
            # fourier_transform_matrix_cache=self.fourier_transform_matrix_cache,
            show_progress=True,
        )

        return transform_matrix @ coeffs
```
