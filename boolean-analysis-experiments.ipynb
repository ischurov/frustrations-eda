```python
eigenstates = 10

lattices = {
    "square4x4": SquareLattice(width=4, height=4),
    "kagome2x2": KagomeLattice(width=2, height=2),
}
J2values = np.arange(0.0, 1.0, 0.1)

analyzers = {
    (J2, lat_name): BooleanFourierAnalyser(
        system=HeisenbergJ1J2(
            lattice,
            J1=1,
            J2=J2,
            use_symmetries=True,
            spin_inversion=1,
        ),
        fourier_transform_matrix_cache=fourier_transform_matrix_cache,
        use_symmetries=True,
        eigenstates=eigenstates,
    )
    for J2 in tqdm(J2values)
    for lat_name, lattice in (lattices.items())
}
```

```python
def get_all_time_top_subsets(analyzers, lat_name, top, eigenstate=0):
    subsets = set([])
    for J2 in tqdm(J2values):
        analyzer = analyzers[J2, lat_name]
        spectre_df = analyzer.get_spectre_df(
            SignalOption(eigenstate=eigenstate)
        )
        subsets.update(spectre_df.iloc[:top].index)
    return subsets
```

```python
len(get_all_time_top_subsets(analyzers, "square4x4", 10))
```

```python
lat_name = "square4x4"
top = 5
eigenstate = 0


def plot_top_state_coeffs(
    analyzers,
    lat_name,
    top,
    eigenstate=0,
    keep_first_accuracy=(1, 2, 5, 10),
    y="abs_coeff",
):
    subsets = list(
        get_all_time_top_subsets(analyzers, lat_name, top, eigenstate)
    )
    data = []
    for J2 in J2values:
        spectre = analyzers[J2, lat_name].get_spectre_df(
            SignalOption(eigenstate=eigenstate)
        )
        data.append(
            spectre.loc[subsets]
            .reset_index()
            .rename(columns={"index": "state"})
            .assign(state=lambda x: x["state"].astype(str))
            .assign(J2=J2)
        )
    df = pd.concat(data, axis=0)
    fig, (ax1, ax2) = plt.subplots(nrows=2, figsize=(10, 8), sharex=True)
    ax1.set_yscale("log")
    sns.lineplot(data=df, x="J2", y=y, hue="state", ax=ax1)

    accuracy_data = []
    for J2 in J2values:
        analyzer = analyzers[J2, lat_name]
        for keep_first in keep_first_accuracy:
            accuracy = accuracy_score(
                np.sign(
                    analyzer.predict(
                        SignalOption("sign", eigenstate), keep_first=keep_first
                    )
                ),
                np.sign(
                    analyzer.system.get_df_eigenstate(
                        k=eigenstate, canonical_basis=True
                    )["eigenstate_coeff"]
                ),
            )
            accuracy_data.append(
                {"J2": J2, "accuracy": accuracy, "keep_first": str(keep_first)}
            )

    sns.lineplot(
        data=pd.DataFrame(accuracy_data),
        x="J2",
        y="accuracy",
        hue="keep_first",
        ax=ax2,
    )
```

```python
plot_top_state_coeffs(
    analyzers,
    "kagome2x2",
    top=5,
    eigenstate=0,
    keep_first_accuracy=[1, 2, 5, 10, 30],
    y="abs_coeff",
)
```

```python
plot_top_state_coeffs(
    analyzers, "square4x4", top=5, eigenstate=0, y="abs_coeff"
)
```
