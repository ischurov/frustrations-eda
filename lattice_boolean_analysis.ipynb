```python
from spin_lattices import KagomeLattice
from heisenberg_hamiltonians import HeisenbergJ1J2
from lattice_boolean_analysis import (
    LBFFromSpinSystem,
    LatticeBooleanAnalyzer,
    SignSignalKind,
    keep_everything,
    AmplitudeSignalKind,
    keep_largest_n,
)
import numpy as np
```

```python
signal = LBFFromSpinSystem(
    HeisenbergJ1J2(
        KagomeLattice(width=2, height=2),
        J1=1.0,
        J2=0.0,
    ),
    eigenstate=0,
    kind=AmplitudeSignalKind()
)
```

```python
analyzer = LatticeBooleanAnalyzer(signal)
```

```python
analyzer.fit()
```

```python
analyzer.truncate(keep_largest_n(10)).prediction_score('neg_mse')
```
