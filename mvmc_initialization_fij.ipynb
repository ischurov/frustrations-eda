```python
from spin_lattices import KagomeLattice, SquareLattice
from heisenberg_hamiltonians import HeisenbergJ1J2
from mvmc_configs import MVMCConfig
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
from itertools import product
import pandas as pd
import seaborn as sns
```

```python
%matplotlib inline
```

```python
system = HeisenbergJ1J2(KagomeLattice(2, 4), J1=1., J2=0., ground_state_cache_dir=Path("groundstates"))
```

```python
# out_dfs = []
for DSROptStepDt, NVMCSample in product([0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9], [1000]):
    mvmc_config = MVMCConfig(
        system,
        total_spin=0,
        NSROptItrStep=500,
        DSROptStepDt=DSROptStepDt,
        NVMCSample=NVMCSample,
        in_orbital_type_to_t={1: 1.0, 2: 0},
    )
    mvmc_config.write_configuration(NVMCCalMode=0)
    mvmc_config.do_monte_carlo_optimization()
    out = mvmc_config.extract_out()
    out_dfs.append(out.assign(DSROptStepDt=DSROptStepDt, NVMCSample=NVMCSample, mvmc_config=mvmc_config))
```
