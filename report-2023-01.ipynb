## Sign structure complexity: Boolean Fourier Analysis perspective

### Objective
We study a multi-body spin system with Heisenberg antiferromagnetic Hamiltonian on the frustrated Kagome lattice. The level of frustrations is determined by the parameter $J_2$ (we put $J_1=1$). It is known that for $J_2 \gtrsim 5.2$, the system is in frustrated regime (spin liquid). We are studying the ground state of the system written in $s_z$ basis. In this case the set of basis elements is a set of sequences $\{\uparrow, \downarrow\}^d$, isomorphic to Boolean hybercube, and the ground state can be seen as a mapping $\{\uparrow, \downarrow\}^d \to \mathbb R$. Such functions can be studied using Boolean Fourier analysis. We will apply it to study the complexity of ground states, their signs and amplitudes, and explain the previous results that involved neural networks.

### What was known
In [Westerhout, et al, 2020](https://doi.org/10.1038/s41467-020-15402-w), it was shown that:

- Neural Networks of various architectures, trained to reconstruct the full ground state of the system after seeing only a small part of it, have different generalization properties depending on whether the system is in frustrated regime or not.
- In unfrustrated regime, the networks generalize well even after seeing only a little bit more than $10^{-3}$ part of the full space. In frustrated regime, even $10^{-2}$ elements are not enough.
- One can decompose the ground state prediction task into predicting amplitudes and signs separately. Amplitudes in the frustrated regime are much easier to learn for NN compared with signs.

### Boolean Fourier Analysis

Consider a space $\mathcal F_d$ of functions $f\colon \{-1, 1\}^d \to \mathbb R$. It is a linear space. Introduce the following scalar product on it:

$$⟨ f | g⟩ := \frac{1}{2^d} \sum_{x \in \{-1, 1\}^d} f(x) g(x).$$

With this scalar product, $\mathcal F_d$ is a real Euclidian (a.k.a. finite-dimensional Hilbert) space.

For any subset $I = \{i_1, i_2, …, i_{k}\} \subset \{1, 2, …, d\}$, denote

$$χ_I(x) := x_{i_1} x_{i_2} ⋯ x_{i_{k}}.$$

Functions $χ_I$ are called _parity functions_. The number of elements in $I$ is called *hamming weight* of $\chi_I$. It is easy to show that the collection of parity functions $(χ_I: I\subset \{1, 2, …, d\})$ is an orthonormal basis in $\mathcal F_d$. Thus any function $f \in \mathcal F_d$ can be uniquely represented as a linear combination of parity functions

$$f = \sum_{I \subset \{1, …, d\}} a_I χ_I,$$

where

$$a_I = ⟨χ_I, f⟩.$$

Coefficients $a_I$ are called Boolean Fourier coefficients of $f$. The set of all Boolean Fourier coefficients is called Boolean Fourier spectre. We will omit Boolean below for brevity.

#### Measures of complexity
##### Terms measures
For each function $f \in \mathcal F_d$, we can take its Fourier decomposition, drop some of the terms in this decomposition and obtain a new function $\tilde f$, which is called a _truncation_ of $f$. To determine, which terms to drop, and which to keep, we have to define _truncation strategy_. We will mostly use `keep_largest_n` truncation strategy, that keeps $n$ largest (by absolute value) coefficients of the Fourier expansion.

Now consider some similarity score $s$ on $\mathcal F_d$, i.e. a symmetric function $s\colon \mathcal F_d\times \mathcal F_d\to \mathbb R$ that measures ”how close” two elements in $\mathcal F_d$ are: the larger value of $s$, the closer elements. For example, overlap is a similarity score. We will also consider sign-accuracy score: $\mathcal{\mathrm{acc}(f, g)}=\#\{x\mid f(x)g(x) > 0\}/2^d$.

For a given similarity score $s$ and a given $f$, one can ask: how many terms we need to keep in the Fourier expansion of $f$ to make a truncation $\tilde f$ close enough to $f$? This value, for a fixed similarity score and a threshold, can be seen as a complexity measure for a boolean function $f$: the more terms we need to approximate it, the more complex it is. We will call this kind of complexity measures *Terms-measures*.

##### Fourier weight and IIPR
Terms measures are computationally expensive: to find the value of such measure for a given function, you have to perform binary search over the number of terms you are keeping. You also have to keep in memory the full Fourier series of the function, that can be intractable for large clusters (number of spins > 30). Thus we consider also much cheaper alternatives: Fourier weight and IIPR complexity measures. They are based on the same idea: simple functions are those whose Fourier series has only a small number of large coefficients.

- Fourier weight of a term is the square of the corresponding coefficient, normalized so that the sum of all weights is 1. Fourier weight complexity measure of a function $f$ under threshold $\alpha$ is the number of terms with the largest weights whose joint weight is at least $\alpha$.

- IPR (inverse participation ratio) is the sum of squares of Fourier weights (i.e. fourth power of coefficients) of all terms. IPR is equal to 1 if there is only one non-zero coefficient, and it decreases as the coefficients “spread over” the series. So, we can use inverse value 1/IPR as a complexity measure. It is called IIPR (Inverse IPR).
