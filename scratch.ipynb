```python
from scipy.special import logsumexp
import numpy as np
import numpy.typing as npt
from scipy.sparse import csr_matrix
```

```python
def logmatmul(M: npt.NDArray | csr_matrix, log_v: npt.NDArray) -> npt.NDArray[np.complex128]:
    """
    Compute log(M @ v) for a matrix M and a vector v. Output is complex.
    """
    if isinstance(M, np.ndarray):
        log_M = np.log(M + 0j)
    elif isinstance(M, csr_matrix):
        log_M = csr_matrix((np.log(M.data + 0j), M.indices, M.indptr), shape=M.shape)
    else:
        raise ValueError(f"Unsupported matrix type {type(M)}")
    return np.asarray(logsumexp(log_M + log_v, axis=1), dtype=np.complex128)
```

```python
# make random csr matrix with 0.5 density
M = csr_matrix(np.random.rand(100, 100) < 0.01, dtype=np.float64)
v = np.random.rand(100)
assert np.allclose(M @ v, np.exp(logmatmul(M, np.log(v.astype(np.complex128)))).real)
```
