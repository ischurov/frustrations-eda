```python
import numpy as np
from scipy.linalg import schur
from scipy.spatial.transform import Rotation as R
```

```python
V = np.array([[1, 0], [0, 1], [0, 0]])
rot_axis = np.array([1, 1, 0])
rotvec = np.pi / 3 * rot_axis / np.linalg.norm(rot_axis)
T = R.from_rotvec(rotvec=rotvec).as_matrix()
```

```python
np.isclose(T.T , T)
```

```python
T @ V
```

```python
A = V.T @ T @ V
```

```python
A
```

```python
M, U = schur(A)
```
